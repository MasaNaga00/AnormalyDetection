# 4. 操作点を実測する

```python
%run tune_c_alpha.py 遅延_run2.csv
print('\n\n--------------')
%run tune_b_minoe.py 遅延_run2.csv
```

```python
import pandas as pd, tune_b_minoe as tb
s = pd.read_csv('scan_b.csv', encoding='utf-8-sig')
tb.top_list(s, min_oe=2.0, alpha=0.005)   # 境界に何が来るかを目視
tb.confirm('遅延_run2.csv', min_oe=2.0, alpha=0.005)  # 本番(2パス)で確認
tb.nb_profile('遅延_run2.csv')            # 沈黙する群の特定
```

tune_c_alpha.py の結果から「C_ALPHA」を単位月最大 が予算に収まる最大の値に、「C_MIN_COUNT」は件数が減らないとき、alphaより先にこちらを5へ。目安：信号Cの取り分を top_n の3〜5件と考え、単位月最大 ≤ 5 あたりを目標に。件数を減らしたいときは、まず C_MIN_COUNT を 3 → 5 で試してください。こちらは並び順に副作用がなく、小件数のノイズだけを落とします。それで足りなければ C_ALPHA を絞る、という順序が安全です。<br>
tune_b_minoe.py の結果から「B_MIN_OE」はO/E分布のP95 + α。top_list で中身を目視して確定、「B_ALPHA」は触らない（効かないため）
1. B_MIN_OE = 2.0 を先に固定（初回の総件数を抑える）
2. C_MIN_COUNT → C_ALPHA の順に信号Cを絞る
3. 実際に run_month.py を回して 検出器 の内訳を見る
4. 枠が余る／足りなければ 2 に戻る
